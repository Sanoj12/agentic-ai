```python
from groq import Groq
```

```python
#create instance
groq = Groq()
```

```python
messages = [{"role":"user" , "content":"what is deep learning?"}]
```

```python
response = groq.chat.completions.create(
      model="openai/gpt-oss-20b",
      messages=messages
)
```

```python
response.choices[0].message.content
```

```python
query = "please provide brainstorm question"
messages = [{"role":"user" , "content":query}]
```

```python
response = groq.chat.completions.create(
      model="openai/gpt-oss-20b",
      messages=messages
)

result = response.choices[0].message.content
print(result)
```

```python
from IPython.display import Markdown ,display

display(Markdown(result))
```

```python
###first step
question = "pick a business area that might be worth exploring for an Agentic ai opportunity??"
messages = [{"role":"user","content":question}]

response = groq.chat.completions.create(
      model="openai/gpt-oss-20b",
      messages=messages
)
business_idea = response.choices[0].message.content

from IPython.display import Markdown ,display

display(Markdown(business_idea))
```

```python
question = "to present a pain-point of Clinical Decision‑Support for Primary Care (P‑AI) -something challenging that might be ripe for an agentic solution?"
messages = [{"role":"user", "content":question}]

response = groq.chat.completions.create(
      model="openai/gpt-oss-20b",
      messages=messages
)

business_idea = response.choices[0].message.content

from IPython.display import Markdown ,display

display(Markdown(business_idea))
```
