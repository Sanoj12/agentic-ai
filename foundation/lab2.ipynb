```python
import os
import json
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown,display
```

```python
load_dotenv(override=True)
```

```python
groq_api_key = os.getenv("GROQ_API_KEY")
deepseek_api_key =os.getenv("DEEPSEEK_API_KEY")

if groq_api_key:
    print(f"Groq api key exist and {groq_api_key[:6]}")
else:
    print("groq_api_key is not exists")

if  deepseek_api_key:
    print(f"deepseek api key exist and {deepseek_api_key[:4]}")

else:
    print("deepseek api key is not exists")
```

```python
request = "please give tips for llm learning"
messages=[{"role":"user","content":request}]
```

```python
from groq import Groq
groq_client = Groq()

response = groq_client.chat.completions.create(
     model="openai/gpt-oss-20b",
     messages=messages,
     temperature=1
)
```

```python
question=response.choices[0].message.content
print(question)
```

```python
from openai import OpenAI

deepseek = OpenAI(api_key=deepseek_api_key,base_url="https://api.deepseek.com")

response = deepseek.chat.completions.create(
    model ="deepseek-chat",
    messages=messages

)
```

```python
###direct http request
import requests

url = "https://api.deepseek.com/v1/chat/completions"
headers = {
    "Authorization": "Bearer YOUR_DEEPSEEK_API_KEY",
    "Content-Type": "application/json"
}

data={
    "model":"deepseek-chat",
    "messages":[messages]
}

response =requests.post(url,headers=headers ,json=data)
print(response)
```

```python
competitors =[]
answers=[]
messages=[{"role":"user" , "content":question1}]
```

```python
model_name="openai/gpt-oss-20b"
response =groq_client.chat.completions.create(model=model_name,messages=messages)

answer= response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```
